```python
import pandas as pd

df = pd.read_csv("old-data/Anvay Close Connections - Connections.csv")

# Only keep the rows where the first column value starts with a period ('.')
# The first column might be unnamed, get it by position
col0 = df.columns[0]
df_period = df[df[col0].astype(str).str.startswith('.')]

# Display the resulting dataframe
len(df_period)
```

```python
df_period = df_period.drop(columns=[col0])
```

```python
df_period.to_csv("data/anvay_connections.csv", index=False)
```
